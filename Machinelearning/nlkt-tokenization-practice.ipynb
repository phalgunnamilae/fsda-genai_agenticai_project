```python
# Stemming

#PorterStemmer

from nltk.stem import PorterStemmer
```

```python
pst = PorterStemmer()

words = [
    "running", "runner", "runs",
    "easily", "fairly",
    "studies", "studying", "studied",
    "happiness", "happy",
    "organization", "organized", "organizing",
    "connection", "connected", "connecting"
]
```

```python
for word in words:
    print(word , 'stem word is', pst.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
from nltk.corpus import wordnet
from nltk import pos_tag, word_tokenize
```

```python
lematizer = WordNetLemmatizer()
words
def get_wordnet_pos(word):
    tag = pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ,
                "N": wordnet.NOUN,
                "V": wordnet.VERB,
                "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)  # Default to noun

for word in words:
    print(word, "→", lematizer.lemmatize(word, get_wordnet_pos(word)))
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
import re
```

```python
text_vector = '''The cat sat on the mat.
                The dog chased the cat.
                The dog barked loudly.
                Cats and dogs are popular pets'''

ps = PorterStemmer()
wordnet  = WordNetLemmatizer()

corpus = []
sent_val = nltk.sent_tokenize(text_vector)
len(sent_val)
```

```python
for i in range(len(sent_val)):
    review = re.sub('[^a-zA-Z]', ' ', sent_val[i])
    review = review.lower()
    review = review.split()
    
    review = [wordnet.lemmatize(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    print(review)
    corpus.append(review)
```

```python
    
# Creating the Bag of Words model 

# Also we called as document matrix 

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_bow = cv.fit_transform(corpus).toarray()
X_bow
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer()
X_tf = tv.fit_transform(corpus).toarray()
X_tf
```
