```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
data_set = pd.read_csv(r"C:\Users\nphal\fsda-genai_agenticai_project\Machinelearning\Churn_Modelling.csv")
data_set
X = data_set.iloc[:,3:-1].values
y = data_set.iloc[:, -1].values
X
```

```python
data_set.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

X[:, 2] = le.fit_transform(X[:, 2])
```

```python
X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y , train_size= 0.2, random_state= 0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# ANN Model Building Architecture

ann = tf.keras.models.Sequential()  # establish layer one after other .. I/P, Hidden layer, o/p layer
```

```python
ann.add(tf.keras.layers.Dense(units=6)) # input layers 
```

```python
ann.add(tf.keras.layers.Dense(units=6, activation= 'relu')) # Hidden layer1
```

```python
#ann.add(tf.keras.layers.Dense(units=5, activation= 'relu')) # Hidden layer2
# We cannot build multilayer hidden layers in our machine. The system will hang. so for multiple layer hidden layers usage we should use Pretrained model 
```

```python
#ann.add(tf.keras.layers.Dense(units=4, activation= 'relu')) # Hidden layer3
```

```python
ann.add(tf.keras.layers.Dense(units=1, activation= 'sigmoid')) # output layer
```

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])
```

```python
ann.fit(X_train, y_train, batch_size= 32, epochs=75)
```
